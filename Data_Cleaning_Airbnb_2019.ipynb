```python
missing_data = df.isnull().sum()

print("Jumlah missing values setiap kolom:")
print(missing_data)

print("\nTotal seluruh missing values:")
print(missing_data.sum())
```

```python
jumlah_duplikat = df.duplicated().sum()

print("Jumlah baris duplikat:", jumlah_duplikat)
```

```python
df_clean = df.copy()

print("Salinan dataset berhasil dibuat!")
print("Jumlah baris awal:", len(df_clean))
```

```python
df_clean = df_clean[df_clean["price"] > 0]

print("Jumlah baris setelah pembersihan:", len(df_clean))
print("Jumlah data harga 0 atau negatif:",
      (df_clean["price"] <= 0).sum())
```

```python
df_clean["name"] = df_clean["name"].fillna("Unknown")

df_clean["host_name"] = df_clean["host_name"].fillna("Unknown")

print("Missing values name:",
      df_clean["name"].isnull().sum())

print("Missing values host_name:",
      df_clean["host_name"].isnull().sum())
```

```python
df_clean["reviews_per_month"] = df_clean["reviews_per_month"].fillna(0)

print("Missing reviews_per_month:",
      df_clean["reviews_per_month"].isnull().sum())

print("Missing last_review:",
      df_clean["last_review"].isnull().sum())
```

```python
print("Jumlah data last_review kosong:",
      df_clean["last_review"].isna().sum())

print("Jumlah data tanpa tanggal ulasan dan tanpa ulasan:",
      (df_clean.loc[df_clean["last_review"].isna(), "number_of_reviews"] == 0).sum())
```

```python
df_clean["last_review"] = pd.to_datetime(
    df_clean["last_review"], errors="coerce"
)

print("Tipe data last_review:",
      df_clean["last_review"].dtype)

print("Jumlah data kosong:",
      df_clean["last_review"].isna().sum())
```

```python
print("Jumlah baris dan kolom:", df_clean.shape)

print("\nJumlah duplikat:",
      df_clean.duplicated().sum())

print("\nJumlah missing values setiap kolom:")
print(df_clean.isnull().sum())

print("\nInformasi dataset:")
df_clean.info()
```

```python
df_clean.to_csv("AB_NYC_2019_clean.csv", index=False)

print("Dataset bersih berhasil disimpan!")
```
